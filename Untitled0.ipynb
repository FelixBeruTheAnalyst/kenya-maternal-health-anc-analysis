```python
import pandas as pd
import matplotlib.pyplot as plt

import io
df = pd.read_csv(io.BytesIO(list(uploaded.values())[0]), encoding='latin1')

df.columns = df.columns.str.strip()
print(df.head())
print(df.columns.tolist())
```

```python
# Sort counties by skilled ANC coverage
df_sorted = df.sort_values('Percentage receiving antenatal care from a skilled provider', ascending=True)

# Set up the chart
fig, ax = plt.subplots(figsize=(10, 14))

# Create horizontal bar chart
bars = ax.barh(
    df_sorted['County'],
    df_sorted['Percentage receiving antenatal care from a skilled provider'],
    color='steelblue',
    edgecolor='white'
)

# Add percentage labels on each bar
for bar in bars:
    width = bar.get_width()
    ax.text(width + 0.5, bar.get_y() + bar.get_height()/2,
            f'{width:.1f}%', va='center', fontsize=8)

# Titles and labels
ax.set_title('Skilled Antenatal Care Coverage by County in Kenya\nKDHS 2022',
             fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Percentage of Women Receiving Skilled ANC (%)', fontsize=11)
ax.set_xlim(0, 110)
ax.axvline(x=df['Percentage receiving antenatal care from a skilled provider'].mean(),
           color='red', linestyle='--', linewidth=1.2, label='National Average')
ax.legend()

plt.tight_layout()
plt.savefig('kenya_anc_coverage.png', dpi=150, bbox_inches='tight')
plt.show()

print("Chart saved!")
```

```python
# Define county categories
low_counties = ['Mandera', 'Garissa', 'Wajir']
high_counties = ['Nairobi City', 'Nyamira', 'Migori']
national_avg = df['Percentage receiving antenatal care from a skilled provider'].mean()

# Assign colors
def get_color(county):
    if county in low_counties:
        return '#e74c3c'  # Red for lowest
    elif county in high_counties:
        return '#2ecc71'  # Green for highest
    else:
        return '#95a5a6'  # Grey for rest

colors = [get_color(c) for c in df_sorted['County']]

# Rebuild chart with colors
fig, ax = plt.subplots(figsize=(10, 14))

bars = ax.barh(
    df_sorted['County'],
    df_sorted['Percentage receiving antenatal care from a skilled provider'],
    color=colors,
    edgecolor='white'
)

# Add percentage labels
for bar in bars:
    width = bar.get_width()
    ax.text(width + 0.5, bar.get_y() + bar.get_height()/2,
            f'{width:.1f}%', va='center', fontsize=8)

# National average line
ax.axvline(x=national_avg, color='navy', linestyle='--',
           linewidth=1.2, label=f'National Average ({national_avg:.1f}%)')

# Titles
ax.set_title('Skilled Antenatal Care Coverage by County in Kenya\nKDHS 2022',
             fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Percentage of Women Receiving Skilled ANC (%)', fontsize=11)
ax.set_xlim(0, 115)

# Legend
from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor='#e74c3c', label='Lowest coverage counties'),
    Patch(facecolor='#2ecc71', label='Highest coverage counties'),
    Patch(facecolor='#95a5a6', label='Other counties'),
    plt.Line2D([0], [0], color='navy', linestyle='--',
               label=f'National Average ({national_avg:.1f}%)')
]
ax.legend(handles=legend_elements, loc='lower right', fontsize=9)

# Caption
fig.text(0.5, -0.01,
         'Source: Kenya Demographic and Health Survey (KDHS) 2022 | Visualization by Felix',
         ha='center', fontsize=8, color='gray')

plt.tight_layout()
plt.savefig('kenya_anc_coverage_final.png', dpi=150, bbox_inches='tight')
plt.show()

print(f"National Average: {national_avg:.1f}%")
```
